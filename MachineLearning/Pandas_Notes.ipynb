# 🐼 Pandas — Complete Beginner's Guide

> **Pandas** = **Pan**el **Da**ta (from econometrics)  
> It is the most popular Python library for **data analysis and manipulation**.  
> If NumPy is for numbers, Pandas is for **tables** (like Excel/Google Sheets but in Python).

| Task | Without Pandas | With Pandas |
|------|---------------|-------------|
| Read a CSV file | `open()`, loops, split | `pd.read_csv('file.csv')` |
| Filter rows | Manual loops | `df[df['marks'] > 90]` |
| Group by category | Dict + loops | `df.groupby('class').mean()` |
| Handle missing data | if-else checks | `df.fillna(0)` |

## 2. Series — 1-D Data

A **Series** is like a single column of data with labels (index).

> 🧠 **Think of Series as:** A numbered (or labeled) list — like one column from an Excel sheet.

````python
# From a list (auto-numbered index)
marks = pd.Series([85, 90, 78, 92, 88])
print("Series from list:")
print(marks)
````

````python
# With custom labels (index)
marks = pd.Series(
    [85, 90, 78, 92, 88],
    index=['Alice', 'Bob', 'Charlie', 'David', 'Eve']
)
print("Series with custom index:")
print(marks)
print()

# Access by label
print(f"Bob's marks: {marks['Bob']}")
print(f"Alice & Eve: \n{marks[['Alice', 'Eve']]}")
````

````python
# From a dictionary
ages = pd.Series({'Alice': 20, 'Bob': 22, 'Charlie': 21})
print("Series from dictionary:")
print(ages)
````

## 3. DataFrame — 2-D Data (Tables)

A **DataFrame** is a table with rows and columns. This is what you'll use 90% of the time!

### 3.1 Creating a DataFrame from a Dictionary (most common)

## 6. Selecting Data

### 6.1 Select Columns

````python
# Single column (returns Series)
print("=== Single column ===")
print(df['Name'])
print()

# Multiple columns (returns DataFrame)
print("=== Multiple columns ===")
print(df[['Name', 'Marks']])
````

## 7. Filtering Data

### 7.1 Basic Filtering

````python
# Students with marks above 85
print("=== Marks > 85 ===")
print(df[df['Marks'] > 85])
print()

# Students from Delhi
print("=== City == Delhi ===")
print(df[df['City'] == 'Delhi'])
````

### 7.2 Multiple Conditions

> ⚠️ **Important:** Use `&` (not `and`) and `|` (not `or`). Always wrap each condition in `()`.

````python
# AND: marks > 85 AND city is Mumbai
print("=== Marks > 85 AND City == Mumbai ===")
print(df[(df['Marks'] > 85) & (df['City'] == 'Mumbai')])
print()

# OR: marks > 90 OR age < 21
print("=== Marks > 90 OR Age < 21 ===")
print(df[(df['Marks'] > 90) | (df['Age'] < 21)])
````

````python
# isin() — check membership
print("=== Age is 20 or 22 ===")
print(df[df['Age'].isin([20, 22])])
````

````python
# query() — cleaner syntax
print("=== Using query() ===")
print(df.query('Marks > 85 and City == "Mumbai"'))
````

## 8. Adding & Removing Columns/Rows

### 8.1 Adding Columns

````python
# Make a copy to experiment
df2 = df.copy()

# New column with a formula
df2['Percentage'] = df2['Marks']   # Already out of 100

# New column based on condition
df2['Result'] = df2['Marks'].apply(lambda x: 'Pass' if x >= 60 else 'Fail')

# New column with a fixed value
df2['School'] = 'ABC High School'

print(df2)
````

### 8.2 Removing Columns

````python
# Drop one column
df2 = df2.drop('School', axis=1)
print("After dropping 'School':")
print(df2.columns.tolist())

# Drop multiple columns
df2 = df2.drop(['Percentage', 'Result'], axis=1)
print(f"After dropping more: {df2.columns.tolist()}")
````

### 8.3 Adding & Removing Rows

````python
# Add a new row
new_student = pd.DataFrame({'Name': ['Henry'], 'Age': [21], 'City': ['Pune'], 'Marks': [82]})
df3 = pd.concat([df2, new_student], ignore_index=True)
print("After adding Henry:")
print(df3.tail(3))
print()

# Remove row by condition (keep only passing students)
df3_pass = df3[df3['Marks'] >= 70]
print(f"After keeping marks >= 70: {len(df3_pass)} students remain")
````

## 9. Handling Missing Data

Missing data is very common in real datasets (empty cells, NaN, null).

````python
# Create sample data with missing values
data_missing = {
    'Name':  ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age':   [20, None, 21, 23, None],
    'Marks': [85, 90, None, 92, 88]
}
df_m = pd.DataFrame(data_missing)
print("Data with missing values (NaN):")
print(df_m)
````

### 9.1 Detecting Missing Values

````python
# Check for missing values
print("=== Missing values (True = NaN) ===")
print(df_m.isnull())
print()

# Count NaN per column
print("=== NaN count per column ===")
print(df_m.isnull().sum())
print()

print(f"Total missing values: {df_m.isnull().sum().sum()}")
````

### 9.2 Handling Missing Values

````python
# Option 1: DROP rows with any missing value
print("=== dropna() — Remove rows with NaN ===")
print(df_m.dropna())
print()
````

````python
# Option 2: FILL missing values with a number
print("=== fillna(0) — Replace NaN with 0 ===")
print(df_m.fillna(0))
print()
````

````python
# Option 3: FILL with mean (for numeric columns)
df_filled = df_m.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Marks'] = df_filled['Marks'].fillna(df_filled['Marks'].mean())
print("=== fillna(mean) — Replace NaN with column mean ===")
print(df_filled)
````

## 11. Grouping & Aggregation

GroupBy is one of Pandas' most powerful features — like pivot tables in Excel!

### 11.1 Basic GroupBy

````python
data_gb = {
    'Name':    ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'City':    ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Subject': ['Math', 'Math', 'Science', 'Math', 'Science', 'Science'],
    'Marks':   [85, 90, 78, 92, 88, 67]
}
df_gb = pd.DataFrame(data_gb)
print("Data:")
print(df_gb)
print()

# Average marks per city
print("=== Average marks per City ===")
print(df_gb.groupby('City')['Marks'].mean())
````

````python
# Multiple aggregations
print("=== Multiple aggregations per City ===")
print(df_gb.groupby('City')['Marks'].agg(['mean', 'min', 'max', 'count']))
````

### 11.2 Group by Multiple Columns

````python
# Average marks per city per subject
print("=== Average marks per City per Subject ===")
print(df_gb.groupby(['City', 'Subject'])['Marks'].mean())
````

### 11.3 Named Aggregations (Clean Output)

````python
result = df_gb.groupby('City').agg(
    avg_marks=('Marks', 'mean'),
    top_marks=('Marks', 'max'),
    student_count=('Name', 'count')
).reset_index()

print("=== Named aggregations ===")
print(result)
````

## 12. Merging & Joining DataFrames

Like SQL JOINs — combine tables based on a common column.

### 12.1 Merge (SQL-style Join)

````python
# Student info
students = pd.DataFrame({
    'ID':   [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

# Student marks
marks_df = pd.DataFrame({
    'ID':    [1, 2, 3, 5],
    'Marks': [85, 90, 78, 95]
})

print("Students table:")
print(students)
print()
print("Marks table:")
print(marks_df)
````

````python
# Inner join (only matching IDs)
print("=== INNER join ===")
print(pd.merge(students, marks_df, on='ID', how='inner'))
print()

# Left join (all students, marks if available)
print("=== LEFT join ===")
print(pd.merge(students, marks_df, on='ID', how='left'))
print()

# Outer join (all from both)
print("=== OUTER join ===")
print(pd.merge(students, marks_df, on='ID', how='outer'))
````

### 12.2 Concat (Stack DataFrames)

````python
# Stack vertically (add more rows)
df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Marks': [85, 90]})
df2 = pd.DataFrame({'Name': ['Charlie', 'David'], 'Marks': [78, 92]})

combined = pd.concat([df1, df2], ignore_index=True)
print("=== Vertical concat ===")
print(combined)
````

## 13. String Operations

Pandas has built-in string methods through `.str` accessor!

````python
df_str = pd.DataFrame({
    'Name':  ['alice smith', 'BOB JONES', 'Charlie Brown', '  david  '],
    'Email': ['alice@gmail.com', 'bob@yahoo.com', 'charlie@gmail.com', 'david@outlook.com']
})

# Case conversion
print("=== upper() ===")
print(df_str['Name'].str.upper())
print()

print("=== title() ===")
print(df_str['Name'].str.title())
print()

print("=== strip() ===")
print(df_str['Name'].str.strip())
````

````python
# Contains (search within strings)
print("=== Email contains 'gmail' ===")
print(df_str['Email'].str.contains('gmail'))
print()

# Extract domain from email
df_str['Domain'] = df_str['Email'].str.split('@').str[1]
print("=== Extracted domains ===")
print(df_str[['Email', 'Domain']])
````

## 14. Apply & Lambda Functions

Apply lets you run **any function** on your data — ultimate flexibility!

### 14.1 Apply with a Named Function

````python
df_apply = pd.DataFrame({
    'Name':  ['Alice', 'Bob', 'Charlie'],
    'Marks': [85, 45, 78]
})

def get_grade(marks):
    if marks >= 90: return 'A+'
    elif marks >= 80: return 'A'
    elif marks >= 70: return 'B'
    elif marks >= 60: return 'C'
    else: return 'F'

df_apply['Grade'] = df_apply['Marks'].apply(get_grade)
print(df_apply)
````

### 14.2 Lambda Functions (quick one-liners)

````python
# Double the marks
df_apply['Double'] = df_apply['Marks'].apply(lambda x: x * 2)

# Pass/Fail
df_apply['Result'] = df_apply['Marks'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

# Apply to entire DataFrame (row-wise)
df_apply['Summary'] = df_apply.apply(
    lambda row: f"{row['Name']} scored {row['Marks']}", axis=1
)
print(df_apply)
````

### 14.3 Map (for simple value replacement)

````python
# Replace values using a dictionary
grade_map = {'A+': 10, 'A': 9, 'B': 8, 'C': 7, 'F': 0}
df_apply['Grade_Points'] = df_apply['Grade'].map(grade_map)
print(df_apply[['Name', 'Grade', 'Grade_Points']])
````

## 15. Pivot Tables

Like Excel pivot tables — summarize data in a cross-tabulation format.

````python
data_pivot = {
    'Student': ['Alice', 'Alice', 'Bob', 'Bob', 'Charlie', 'Charlie'],
    'Subject': ['Math', 'Science', 'Math', 'Science', 'Math', 'Science'],
    'Marks':   [85, 90, 78, 82, 92, 88]
}
df_pivot = pd.DataFrame(data_pivot)
print("Original data:")
print(df_pivot)
print()

# Pivot table
pivot = df_pivot.pivot_table(
    values='Marks',       # What to aggregate
    index='Student',      # Rows
    columns='Subject',    # Columns
    aggfunc='mean'        # Aggregation function
)
print("Pivot table:")
print(pivot)
````

## 16. Date & Time

````python
# Create date column
df_date = pd.DataFrame({
    'Date': ['2024-01-15', '2024-02-20', '2024-03-25', '2024-04-10'],
    'Sales': [150, 200, 180, 220]
})

# Convert string to datetime
df_date['Date'] = pd.to_datetime(df_date['Date'])

# Extract parts
df_date['Year'] = df_date['Date'].dt.year
df_date['Month'] = df_date['Date'].dt.month
df_date['Day'] = df_date['Date'].dt.day
df_date['Day_Name'] = df_date['Date'].dt.day_name()
df_date['Quarter'] = df_date['Date'].dt.quarter

print(df_date)
````

````python
# Filter by date range
mask = (df_date['Date'] >= '2024-02-01') & (df_date['Date'] <= '2024-03-31')
print("=== Feb to March data ===")
print(df_date[mask])
````

## 17. 🎯 Real-World Mini Project: Student Report Card

````python
import pandas as pd

# Student data
data = {
    'Name':    ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Math':    [85, 90, 78, 45, 92],
    'Science': [90, 88, 72, 55, 95],
    'English': [78, 95, 80, 60, 88],
    'History': [88, 82, 76, 40, 91]
}
df_report = pd.DataFrame(data)

# Calculate total and average
subjects = ['Math', 'Science', 'English', 'History']
df_report['Total'] = df_report[subjects].sum(axis=1)
df_report['Average'] = df_report[subjects].mean(axis=1).round(2)

# Assign grades
def assign_grade(avg):
    if avg >= 90: return 'A+'
    elif avg >= 80: return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 60: return 'C'
    else: return 'F'

df_report['Grade'] = df_report['Average'].apply(assign_grade)

# Assign rank
df_report['Rank'] = df_report['Average'].rank(ascending=False).astype(int)

# Sort by rank
df_report = df_report.sort_values('Rank')

print("📋 STUDENT REPORT CARD")
print("=" * 70)
print(df_report.to_string(index=False))
print("=" * 70)
print(f"\n🏆 Topper: {df_report.iloc[0]['Name']} with {df_report.iloc[0]['Average']}% average!")
````

## 18. 🎯 Mini Project 2: Sales Data Analysis

````python
import pandas as pd
import numpy as np

# Create sample sales data
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
data_sales = {
    'Date': dates,
    'Product': np.random.choice(['Laptop', 'Phone', 'Tablet', 'Headphones'], 100),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'Units': np.random.randint(1, 20, 100),
    'Price': np.random.choice([999, 699, 499, 149], 100)
}
df_sales = pd.DataFrame(data_sales)
df_sales['Revenue'] = df_sales['Units'] * df_sales['Price']

print("Sample data:")
print(df_sales.head())
print(f"\nTotal rows: {len(df_sales)}")
````

````python
# Revenue by Product
print("📊 Revenue by Product:")
print(df_sales.groupby('Product')['Revenue'].sum().sort_values(ascending=False))
print()

# Revenue by Region
print("📊 Revenue by Region:")
print(df_sales.groupby('Region')['Revenue'].sum().sort_values(ascending=False))
````

## 📝 Quick Cheat Sheet

```
CREATION:
  pd.Series([1,2,3])                → 1-D data
  pd.DataFrame({'col': [1,2,3]})    → 2-D table
  pd.read_csv('file.csv')           → from CSV

INSPECTION:
  df.head()  df.tail()  df.shape  df.info()  df.describe()

SELECTION:
  df['col']           → single column
  df.iloc[0]          → by position
  df.loc[0, 'col']    → by label

FILTERING:
  df[df['col'] > 5]                     → condition
  df[(df['A'] > 5) & (df['B'] < 10)]   → multiple

MISSING DATA:
  df.isnull().sum()  df.dropna()  df.fillna(value)

GROUPING:
  df.groupby('col').mean()
  df.groupby('col').agg(['mean','sum'])

MERGING:
  pd.merge(df1, df2, on='key')  pd.concat([df1, df2])
```
